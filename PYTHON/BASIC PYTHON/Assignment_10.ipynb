**3. Write a function intersection(list1, list2) that returns common elements between two lists without
duplicates.**

```python
def intersection(list1, list2):
    result = []
    for i in list1:
        if i in list2 and i not in result:
            result.append(i)
    return result
list1 = list(map(int, input().split()))
list2 = list(map(int, input().split()))
print(intersection(list1, list2))
```

**5. Write a function that takes a list of integers and a target value. Return the indices of two
numbers whose sum equals the target.**

```python
def two_sum(numbers, target):
    for i in range(len(numbers)):
        for j in range(i + 1, len(numbers)):
            if numbers[i] + numbers[j] == target:
                return [i, j]
numbers = list(map(int, input().split()))
target = int(input())
print(two_sum(numbers, target))
```

**1. Write a function generate_password(length=8) that generates a random secure password of given
length.**

```python
import random
import string
def generate_password(length=8):
    characters = string.ascii_letters + string.digits + string.punctuation
    password = ''
    for i in range(length):
        password = password + random.choice(characters)
    return password
length = int(input())
print(generate_password(length))
```

**5. Write a function validate_password (password, min_length=8) that checks password rules
(uppercase, lowercase, number, special character).**

```python
def validate_password(password, min_length=8):
    if len(password) >= min_length and any(i.isupper() for i in password) and any(i.islower() for i in password) and any(i.isdigit() for i in password) and any(not i.isalnum() for i in password):
        return True
    else:
        return False
password = input('Enter password: ')
print(validate_password(password))
```

**2. Write a function using args that returns all prime numbers among the arguments.**

```python
def prime_numbers(*args):
    result = []
    for n in args:
        if n > 1:
            for i in range(2, n):
                if n % i == 0:
                    break
            else:
                result.append(n)
    return result
numbers = list(map(int, input('Enter numbers: ').split()))
print(prime_numbers(*numbers))
```

**3. Write a function that returns True if a number is an Armstrong number, otherwise False.**

```python
def is_armstrong(n):
    temp = n
    total = 0
    power = len(str(n))
    while temp > 0:
        digit = temp % 10
        total = total + digit ** power
        temp = temp // 10
    return total == n
n = int(input('Enter a number: '))
print(is_armstrong(n))
```

```python
def is_armstrong(n):
    temp = n
    total = 0
    power = len(str(n))
    while temp > 0:
        digit = temp % 10
        total = total + digit ** power
        temp = temp // 10
    return total == n
n = int(input('Enter a number: '))
print(is_armstrong(n))
```

**5. Write a function that returns the number of uppercase, lowercase, and digits in a string.**

```python
def count_characters(s):
    upper = 0
    lower = 0
    digits = 0
    for i in s:
        if i.isupper():
            upper += 1
        elif i.islower():
            lower += 1
        elif i.isdigit():
            digits += 1
    return upper, lower, digits
s = input('Enter a string: ')
print(count_characters(s))
```
