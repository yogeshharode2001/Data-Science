**1. Ramesh’s basic salary is input through the keyboard. His dearness allowance is 40% of basic salary, and house rent allowance is 20% of basic salary. Write a program to calculate his gross salary.**

```python
basic_salary = eval(input('Enter basic salary :'))
dearness_allowance = basic_salary * 40 / 100
house_rent_allowance = basic_salary * 20 / 100
gross_salary = basic_salary + dearness_allowance + house_rent_allowance
print(f'Ramesh is {basic_salary} His {dearness_allowance} and {house_rent_allowance} Calculated His Gross Salary is : {gross_salary}')
```

**2. The distance between two cities (in km.) is input through the keyboard. Write a program to convert and print this distance in meters, feet, inches and centimeters.**

```python
km = eval(input('Enter distance in kilometers :'))
meters = km * 1000
feet = km * 3280.84
inches = km * 39370.08
centimeters = km * 100000
print(f'Distance in meters : {meters}')
print(f'Distance in feet : {feet}')
print(f'Distance in inches : {inches}')
print(f'Distance in centimeters : {centimeters}')
```

**3. If the marks obtained by a student in five different subjects are input through the keyboard, find out the aggregate marks and percentage marks obtained by the student. Assume that the maximum marks that can be obtained by a student in each subject is 100.**

```python
m1 = eval(input('Enter marks in subject 1 :'))
m2 = eval(input('Enter marks in subject 2 :'))
m3 = eval(input('Enter marks in subject 3 :'))
m4 = eval(input('Enter marks in subject 4 :'))
m5 = eval(input('Enter marks in subject 5 :'))
aggregate_marks = m1 + m2 + m3 + m4 + m5
percentage_marks = aggregate_marks / 500 * 100
print(f'Aggregate Marks is : {aggregate_marks}')
print(f'Percentage Marks is : {percentage_marks}')
```

**4. Temperature of a city in Fahrenheit degrees is input through the keyboard. Write a program to convert this temperature into Centigrade degrees.**

```python
fahrenheit = eval(input("Enter temperature in Fahrenheit: "))
centigrade = (fahrenheit - 32) * 5/9
print(f'That converts a Temperature from {fahrenheit} Fahrenheit to Centigrade Degrees is : {centigrade}')
```

**5. The length & breadth of a rectangle and radius of a circle are input through the keyboard. Write a program to calculate the area & perimeter of the rectangle, and the area & circumference of the circle.**

```python
l = eval(input('Enter a length of ractangle :'))
w = eval(input('Enter a width of ractangle :'))
r = eval(input('Enter a radius of circle :'))
pi = 3.14
area_of_ractangle = l * w
perimeter_of_ractangle = 2*(l + w)
area_of_circle = pi * r * r
circumference_of_a_circle = 2 * pi * r
print(f'Area of ractangle {l} or {w} is :{area_of_ractangle} ')
print(f'Perimeter of a Rectangle {l} or {w} is : {perimeter_of_ractangle}')
print(f'Area of a Circle is : {area_of_circle}')
print(f'Circumference of a Circle is : {circumference_of_a_circle}')
```

**7. If a five-digit number is input through the keyboard, write a program to reverse the number.**

```python
num = eval(input('Enter a five-digit number :'))
a = num % 10
b = (num // 10) % 10
c = (num // 100) % 10
d = (num // 1000) % 10
e = num // 10000
reverse_number = a * 10000 + b * 1000 + c * 100 + d * 10 + e
print(f'Reverse the Number is : {reverse_number}')
```

**9. In a town, the percentage of men is 52. The percentage of total literacy is 48. If total percentage of literate men is 35 of the total population, write a program to find the total number of illiterate men and women if the population of the town is 80,000.**

```python
population = 80000

men = population * 52 / 100
women = population - men

literate = population * 48 / 100
literate_men = population * 35 / 100
literate_women = literate - literate_men

illiterate_men = men - literate_men
illiterate_women = women - literate_women

print(f'The Total Number of Illiterate Men is : {illiterate_men}')
print(f'The Total Number of Illiterate Women is : {illiterate_women}')
```
