**8. Simple Interest**
    **Question:-**
        -  Write a Python program that accepts principal amount, rate of interest, and time period and
        -  calculates simple interest.

```python
# P = Principal (amount invested or borrowed)
# R = Rate of interest per year (%)
# T = Time (in years)

# Simple Interest=(P×R×T)/100
```

```python
p = eval(input('Enter a Principal amount :'))
r = eval(input('Enter a Rate of Interest :'))
t = eval(input('Enter a Time Period of :'))
simple_interest = (p * r * t)/100
print(f'Calculated Simple Interest is : {simple_interest}')
```

**9. Total Amount Using Simple Interest**
    **Question:-**
        -  Write a Python program that calculates the final amount after adding simple interest to the
        -  principal.

```python
# Total Amount = P + SI
# P = Principal (amount invested or borrowed)
# R = Rate of interest per year (%)
# T = Time (in years)

# Simple Interest=(P×R×T)/100
```

```python
p = eval(input('Enter a Principal amount :'))
r = eval(input('Enter a Rate of Interest :'))
t = eval(input('Enter a Time Period of :'))
simple_interest = (p * r * t)/100
total_amount = p + simple_interest
print(f'Calculated Simple Interest is : {simple_interest}')
print(f'Calculated Total Amount is : {total_amount}')
```

**12. Celsius to Fahrenheit**
    **Question:-**
        -  Write a Python program that converts a temperature from Celsius to Fahrenheit.

```python
# Celsius to Fahrenheit

# F=(C*9/5)+32

# Fahrenheit to Celsius

# C=(F−32)*5/9
```

```python
celsius = eval(input("Enter temperature in Celsius: "))
fahrenheit = (celsius * 9/5) + 32
print(f'That converts a Temperature from {celsius} Celsius to Fahrenheit is : {fahrenheit}')
```

**13. Fahrenheit to Celsius**

**17. Volume of a Cylinder**
    **Question:-**
        -  Write a Python program that accepts the radius and height of a cylinder and calculates its volume.

```python
# V = Volume
# π (pi) ≈ 3.1416
# r = Radius of the circular base
# h = Height of the cylinder

# The formula for the volume of a cylinder is:
#                                             V = π * r^2 * h
```

```python
r = eval(input('Enter a Radius of the circular base :'))
h = eval(input('Enter a Height of the cylinder :'))
pi = 3.14
volume_of_a_cylinder = pi * r * r * h
print(f'Calculated Volume of a Cylinder is : {volume_of_a_cylinder}')
```

**18. Distance Between Two Points**
    **Question:-**
        -  Write a Python program that accepts coordinates ((x_1,y_1)) and ((x_2,y_2)) and calculates the
        -  distance between them.

```python
x1 = eval(input('Enter x1 :'))
y1 = eval(input('Enter y1 :'))
x2 = eval(input('Enter x2 :'))
y2 = eval(input('Enter y2 :'))
distance_between_two_points = ((x2 - x1)**2 + (y2 - y1)**2)**0.5
print(f'Distance Between Two Points is : {distance_between_two_points}')
```
