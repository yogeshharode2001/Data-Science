```python
i = 0
while i < 3:
    print(i)
    i = i + 1

# Step-1:   i = 0   while 0 < 3   True   i = 0 + 1    i = 1
# Step-1:   i = 1   while 1 < 3   True   i = 1 + 1    i = 2
# Step-1:   i = 2   while 2 < 3   True   i = 2 + 1    i = 3
# Step-1:   i = 3   while 3 < 3   False   
```

```python
i = 1
while i < 0:
    print(i)
    i = i + 1

# Step-1:   i = 1   while 1 < 0   True   i = 1 + 1    i = 2
# Step-1:   i = 2   while 1 < 0   True   i = 2 + 1    i = 3
# Step-1:   i = 3   while 2 < 0   True   i = 3 + 1    i = 4
# Step-1:   i = 4   while 3 < 0   True   i = 4 + 1    i = 5
```

```python
i = 1
while i != 0:
    print(i)
    i = i + 1

# Step-1:   i = 1   while 1 < 0   True  print(1)   i = 1 + 1    i = 2
# Step-1:   i = 2   while 1 < 0   True  print(2)   i = 2 + 1    i = 3
# Step-1:   i = 3   while 2 < 0   True  print(3)   i = 3 + 1    i = 4
# Step-1:   i = 4   while 3 < 0   True  print(4)   i = 4 + 1    i = 5
```
