**Format**

```python
n1 = 100
n2 = 200
add = n1 + n2

print('the addition of 10 and 20 is equal to 30')
print('the addition of {} and {} is equal to {}')
print('the addition of {} and {} is equal to {}'.format(n1,n2,add))


print('the addition of 10 and 20 = 30')
print('the addition of {} and {} = {}'.format(n1,n2,add))


print('10+20= 30')
print('{}+{}={}'.format(n1,n2,add))


print('the summation of 10,20:30')
print('the summation of {},{}:{}'.format(n1,n2,add))
```

**Fstring**

```python
n1 = 100
n2 = 200
add = n1 + n2

print('the addition of 10 and 20 is equal to 30')
print('the addition of {} and {} is equal to {}')
print('the addition of {n1} and {n2} is equal to {add}')
print(f'the addition of {n1} and {n2} is equal to {add}')


print(f"the addition of {n1} and {n2} = {add}")
print(f"{n1}+{n2}={add}")
print(f"the summation of {n1},{n2}:{add}")
```
