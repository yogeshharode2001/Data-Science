**Ask the enter a number and find the square of the number reapeat the above task 5 times**

```python
for i in range(5):
    n = int(input('Enter a number: '))
    print('Square =', n * n)
```

**Ask the user enter a number get it is an even or odd repeat the above task 5 times**

```python
for i in range(5):
    n = int(input('Enter a number: '))
    if n % 2 == 0:
        print('Even')
    else:
        print('Odd')
```

**Ask the user print only 'a' from a string 'hai how are you'**

**Ask the user get a random number between 1 to 10
         user enter a number between 1 to 10
         if both are match print won otheriwse lost
         give only 3 chances**

```python
import random
r = random.randint(1, 10)
for i in range(3):
    n = int(input('Enter a number between 1 to 10: '))
    if n == r:
        print('Won')
        break
    else:
        print('Lost')
```

**Ask the user print only vowels from a  string 'ola ola ola'**
