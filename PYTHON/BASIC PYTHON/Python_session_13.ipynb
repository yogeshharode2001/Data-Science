**Lambda Functions**

- Lambda functions are the functions write in a sigle line

- For the faster run will go with lambda functions approch

```python
def add(x):
    print(x + 10)
add(100)
```

```python
# What is your argument
# What is your output
lambda <argument>:<output>
```

```python
Sum = lambda x : x + 10
Sum(100)
```

```python
Sum = lambda x : x + 10
Sum(1000)
```

```python
def Cube(a):
    print(a**3)
Cube(10)
```

```python
Cube = lambda a : a ** 3
Cube(10)
```

```python
Cube = lambda a : a ** 3
Cube(20)
```

```python
def f1(a,b):
    print(a+b)
f1(10,20)
```

```python
f2 = lambda a,b : a + b
f2(10,20)
```

```python
lambda <argument1>,<argument2>:<output>
```

```python
def avg(a,b,c=30):
    print(round(a + b + c)/3)
avg(10,20)
```

```python
avg = lambda a, b , c = 30 : (round(a + b + c)/3)
avg(10,20)
```

```python
avg = lambda a, b , c = 20 : (round(a + b + c)/3)
avg(10,20)
```

**Lambda functions using list and for loop**

```python
l1 = [21,35,68,98,77]
# I want to multiply each number by 10
l1 * 10
```

```python
l1 = [21,35,68,98,77]
out = []
for i in l1:
    out.append(i * 10)
out
```

```python
l1 = [21,35,68,98,77]
out = [i * 10 for i in l1]
out
```

```python
l1 = [21,35,68,98,77]
lambda i : i + 10
lambda i : i + 10 , l1
map(lambda i : i + 10 , l1)
list(map(lambda i : i + 10 , l1))
```

```python
l1 = [21,35,68,98,77]
lambda i : i * 10
lambda i : i * 10 , l1
map(lambda i : i * 10 , l1)
list(map(lambda i : i * 10 , l1))
```

```python
# l1 = ['hyd','mumbai','pune','blr']
# out = ['HYD','MUMBAI','PUNE','BLR']  # UPPER

# l1 = ['hyd','mumbai','pune','blr']
# out = [0,1,2,3]   # INDEX

# l1 = ['hyd','mumbai','pune','blr']
# out = ['mumbai','pune']

# l1 = ['hyd','mum#bai','pu#ne','blr']
# out = ['hyd','blr']
```

```python
l1 = ['hyd','mumbai','pune','blr']
# out = ['HYD','MUMBAI','PUNE','BLR']  # UPPER
list(map(lambda i : i.upper(),l1))
```

```python
l1 = ['hyd','mumbai','pune','blr']
# out = [0,1,2,3]   # INDEX
list(map(lambda i : l1.index(i),l1))
```

```python
l1 = ['hyd','mumbai','pune','blr']
# out = ['mumbai','pune']
list(map(lambda i : len(i) > 3 , l1))
```

```python
l1 = ['hyd','mumbai','pune','blr']
# out = ['mumbai','pune']
list(filter(lambda i : len(i) > 3 , l1))
```

```python
l1 = ['hyd','mum#bai','pu#ne','blr']
out = ['hyd','blr']
list(map(lambda i : len(i) <= 3 , l1))
```

```python
l1 = ['hyd','mum#bai','pu#ne','blr']
out = ['hyd','blr']
list(filter(lambda i : len(i) <= 3 , l1))
```

```python
l1 = ['hyd','mum#bai','pu#ne','blr']
out = ['hyd','blr']
list(filter(lambda i : '#' not in i , l1))
```

```python
l1 = ['hyd','mum#bai','pu#ne','blr']
out = ['hyd','blr']
list(map(lambda i : '#' not in i , l1))
```
