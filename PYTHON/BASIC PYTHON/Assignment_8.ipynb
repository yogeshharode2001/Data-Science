- Problem Statement
    - Write a program that demonstrates the difference between global and local variables. Define a globalvariable score = 100. Inside a function 
    - update_score(value), use the global score, add a local bonus = 50, and compute new_score. Show that bonus is not accessible outside the function.

- Explanation
    - Variables declared inside a function are local and exist only within that function. Variables declaredoutside are global and can be accessed 
    - (but not modified without 'global' keyword) from inside functions.

- Function Signature
    - score = 100
    - def update_score(value):   # your code here

- Example 1
    - Input:
        - update_score(25)
    - Output:
        - Inside function: new_score = 175
        - Inside function: bonus = 50

- Example 2
    - Input:
        - print("Outside: score =", score)
    - Output:
        - Outside: score = 100

- Example 3
    - Input:
        - print("Outside: bonus =", bonus)
    - Output:
        - NameError: name 'bonus' is not defined

    - Function Type: Demonstrates global vs local variable scope

    - Note: Show the full program and all outputs including the NameError.

```python
score = 100
def update_score(value):
    global score
    bonus = 50
    new_score = score + value + bonus
    print('Inside function: new_score =', new_score)
    print('Inside function: bonus =', bonus)
update_score(25)
print('Outside: score =', score)
try:
    print('Outside: bonus =', bonus)
except NameError as e:
    print('NameError:', e)
```

- Function Signature
    - def safe_divide(a, b):    # your code here

- Example 1
    - Input:
        - safe_divide(10, 2)
    - Output:
        - 5.0

**Q15. [MEDIUM] Percentage Calculator Using Functions**

- Problem Statement
    - A student scores marks in 5 subjects (each out of 100). Write TWO functions:calculate_percentage(marks) that returns the percentage, and assign_grade
    - (percentage) that returns a grade based on the percentage.

- Explanation
    - First calculate the total marks and percentage. Then use if/elif/else to assign a grade based on percentage ranges.

- Function Signature
    - def calculate_percentage(marks):    # your code here

    - def assign_grade(percentage):       # your code here

- Example 1
    - Input:
        - marks = [85, 92, 78, 88, 95]
    - Output:
        - Total Marks = 438 / 500
        - Percentage = 87.6%
        - Grade = A

    - Function Type: Functions: calculate_percentage(marks) and assign_grade(percentage)

    - Note: Grade rules: >=90 -> A+, >=80 -> A, >=70 -> B, >=60 -> C, >=50 -> D, <50 -> Fail

```python
def calculate_percentage(marks):
    total = sum(marks)
    percentage = (total / 500) * 100
    return total, percentage
def assign_grade(percentage):
    if percentage >= 90:
        return 'A+'
    elif percentage >= 80:
        return 'A'
    elif percentage >= 70:
        return 'B'
    elif percentage >= 60:
        return 'C'
    elif percentage >= 50:
        return 'D'
    else:
        return 'Fail'
marks = list(map(int, input().split()))
total, percentage = calculate_percentage(marks)
grade = assign_grade(percentage)
print('Total Marks =', total, '/ 500')
print('Percentage =', percentage, '%')
print('Grade =', grade)
```

**Q18. [MEDIUM] Simple & Compound Interest**

- Problem Statement
    - Write a function interest(principal, rate, time) that returns BOTH Simple Interest and Compound Interest. No loops or recursion needed.

- Explanation
    - Simple Interest = (P * R * T) / 100. Compound Interest = P * ((1 + R/100)^T) - P. Both can be computed directly using arithmetic operations.

- Function Signature
    - def interest(principal, rate, time):     # your code here

- Example 1
    - Input:
        - interest(10000, 5, 3)
    - Output:
        - Simple Interest = 1500.0, Compound Interest = 1576.25

- Example 2
    - Input:
        - interest(5000, 7.5, 2)
    - Output:
        - Simple Interest = 750.0, Compound Interest = 778.13

- Example 3
    - Input:
        - interest(20000, 4.5, 5)
    - Output:
        - Simple Interest = 4500.0, Compound Interest = 4920.37

    - Function Type: Function WITH parameters: principal, rate, time - no loops, no recursion

    - Note: Return both values.

```python
def interest(principal, rate, time):
    si = (principal * rate * time) / 100
    ci = principal * (1 + rate / 100) ** time - principal
    return round(si, 2), round(ci, 2)
p = float(input())
r = float(input())
t = float(input())
si, ci = interest(p, r, t)
print('Simple Interest =', si)
print('Compound Interest =', ci)
```

**Q19. [MEDIUM] Armstrong Number Check (No Loops, No Recursion)**

- Problem Statement
    - Write a function is_armstrong(n) that checks whether n equals the sum of its own digits, each raised to the power of the digit-count. Do NOT use 
    - while, for, or recursion.

- Explanation
    - Convert the number to a string to access digits. Use len() for digit count and sum() with a generator expression to compute the sum of powered 
    - digits. Compare with the original number.

- Function Signature
    - def is_armstrong(n):        # your code here

- Example 1
    - Input:
        - n = 153
    - Output:
        - True (1^3 + 5^3 + 3^3 = 153)

- Example 2
    - Input:

    - Output:
        - True (3^3 + 7^3 + 0^3 = 370)

- Example 3
    - Input:
        - n = 9474

- Example 4
    - Input:
        - n = 123
    - Output:
        - False

    - Function Type: Function WITH parameter: n (integer) - no loops, no recursion

    - Note: Use string conversion + sum() generator expression.

```python
def is_armstrong(n):
    digits = str(n)
    power = len(digits)
    total = sum(int(digit) ** power for digit in digits)
    return total == n
n = int(input())
print(is_armstrong(n))
```
