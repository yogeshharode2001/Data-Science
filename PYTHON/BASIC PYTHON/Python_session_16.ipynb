**Try-Except**

- Try block is use to run the code  when the code is having any error 

    - It will catch by Except block

- If the code is correct then it will execyte as it is

- Except block activate when we have error in try block

- This mechanism called as **Exception Handling**

- In this process we are not stopping the interpreter even we got error

```python
a = 10
b = 20
c = a + d
print('Hi')
```

```python
try:
    a = 10
    b = 20
    print(a)
    print(b)
    c = a + d
except Exception as e :
    print(e)
```
