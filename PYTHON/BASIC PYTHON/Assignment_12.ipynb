**Q2) [2,3,30,40,50,6,7]

**Q3) [2,3,30,40,50,6,7]
     sort the values with out sort method**

```python
a = [2, 3, 30, 40, 50, 6, 7]
for i in range(len(a)):
    for j in range(i + 1, len(a)):
        if a[i] > a[j]:
            a[i], a[j] = a[j], a[i]
print(a)
```

```python
a = [2, 3, 30, 40, 50, 6, 7]
i = 0
while i < len(a):
    j = 0
    while j < len(a)-1:
        if a[j] > a[j+1]:
            a[j], a[j+1] = a[j+1], a[j]
        j = j + 1
    i = i + 1
print(a)
```

**# Q5) [2,3,[30,40,50],[6,7]] output will be = [2,3,30,40,50,6,7]**

```python
a = [2, 3, [30, 40, 50], [6, 7]]
b = []
for i in a:
    if type(i) == list:
        for j in i:
            b.append(j)
    else:
        b.append(i)
print(b)
```

```python
a = [2, 3, [30, 40, 50], [6, 7]]
b = []
for i in a:
    if isinstance(i, list):
        b.extend(i)
    else:
        b.append(i)
print(b)
```

```python
a = [2, 3, [30, 40, 50], [6, 7]]
b = []
i = 0
while i < len(a):
    if isinstance(a[i], list):
        j = 0
        while j < len(a[i]):
            b.append(a[i][j])
            j = j + 1
    else:
        b.append(a[i])
    i = i + 1
print(b)
```

**Q8) str1='hello how are you, hello im good'
     take any sentence like above find the most reptaed word**

```python
str1 = 'hello how are you, hello im good'
a = str1.replace(',', '').split()
word = ''
maximum = 0
for i in a:
    c = 0
    for j in a:
        if i == j:
            c = c + 1
    if c > maximum:
        maximum = c
        word = i
print(word)
```

```python
str1 = 'hello how are you, hello im good'
a = str1.replace(',', '').split()
d = {}
for i in a:
    if i in d:
        d[i] = d[i] + 1
    else:
        d[i] = 1
word = ''
for i in d:
    if d[i] > d.get(word, 0):
        word = i
print(word)
```

```python
str1 = 'hello how are you, hello im good'
a = str1.replace(',', '').split()
d = {}
i = 0
while i < len(a):
    if a[i] in d:
        d[a[i]] = d[a[i]] + 1
    else:
        d[a[i]] = 1
    i = i + 1
word = ''
maximum = 0
for i in d:
    if d[i] > maximum:
        maximum = d[i]
        word = i
print(word)
```

**Q11) create counter program :
      hello=2 how =1 are =2 you =1 im =1 good=1 ,=1
      output should be alphabetical order**

```python
str1 = 'hello hello how are are you im good ,'
a = str1.split()
d = {}
for i in a:
    if i in d:
        d[i] = d[i] + 1
    else:
        d[i] = 1
for i in sorted(d):
    print(i, '=', d[i],end= ' ')
```

```python
str1 = 'hello hello how are are you im good ,'
a = str1.split()
d = {}
i = 0
while i < len(a):
    if a[i] in d:
        d[a[i]] = d[a[i]] + 1
    else:
        d[a[i]] = 1
    i = i + 1
keys = sorted(d)
i = 0
while i < len(keys):
    print(keys[i], '=', d[keys[i]],end= ' ')
    i = i + 1
```

**Q12) str1='The India is an Greatest Country!, Im ha$$$py here&&&'
   out='India Greatest Country,Im hapy here'**

```python
str1 = 'The India is an Greatest Country!, Im ha$$$py here&&&'
out = ''
for i in str1:
    if i.isalpha() or i == ' ' or i == ',':
        out = out + i
a = out.split()
b = []
for i in a:
    if i not in ['The', 'is', 'an']:
        b.append(i)
print(' '.join(b).replace('Country Im', 'Country,Im'))
```

**Q13) [202,301,405,607,590,320,140,50]   T
     - [102,201,305,407,390,220,140,50]   T
     - [102,201,305,407,390,220,440,50]  False
     - you need to create a function
     - the function should verify the above pattern with any list**

```python
def check_pattern(a, b):
    for i in range(len(a)):
        if abs(a[i] - b[i]) not in [0, 100]:
            return False
    return True
a = [202, 301, 405, 607, 590, 320, 140, 50]
b = [102, 201, 305, 407, 390, 220, 140, 50]
print(check_pattern(a, b))
c = [102, 201, 305, 407, 390, 220, 440, 50]
print(check_pattern(b, c))
```

```python
def check_pattern(a, b):
    i = 0
    while i < len(a):
        if a[i] == b[i] or a[i] == b[i] + 100 or b[i] == a[i] + 100:
            i = i + 1
        else:
            return False
    return True
a = [202, 301, 405, 607, 590, 320, 140, 50]
b = [102, 201, 305, 407, 390, 220, 140, 50]
print(check_pattern(a, b))
```
