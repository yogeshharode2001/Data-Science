**List Comprehension**

- List Comprehension are use to write and execute list operations are fast way

- List Comprehension are writing list operations in a single line

```python
l = []
for i in range(1,11):
    l.append(i)
l
```

```python
# your output input
# output : i
# for i in range(1,11)
# L = [output forloop]
l = [i for i in range(1,11)]
l
```

```python
l = [11,14,25,32]
out = []
for i in l:
    out.append(i*i)
out
```

```python
l = [11,14,25,32]
out = [i * i for i in l]
out
```

```python
l = ['hyd','pune','mumbai','blr']
# out = ['HYD','PUNE','MUMBAI','BLR']
out = [i.upper() for i in l]
out    
```

```python
l = [20,24,33,46,78]
out = []
for i in l:
    if i % 2 == 0:
        out.append(i)
out
```

```python
# [<output> <forloop> <ifcodition>]
out = [i for i in l if i % 2 == 0]
out
```

```python
# l = ['hyd','pune','mumbai','blr']
# out = ['mumbai','pune']         # len(i) > 3

# l = ['hyd','pu#ne','mum#bai','blr']
# out = ['hyd','blr']  #  '#' not in i
```

```python
l = ['hyd','pune','mumbai','blr']
out = [i for i in l if len(i) > 3]
out
```

```python
l = ['hyd','pu#ne','mum#bai','blr']
out = [i for i in l if '#' not in i]
out
```

**Case3: For loop with if-else**

```python
# case-1: [<output> <forloop>]
# case-2: [<output> <forloop> <ifcodition>]
# case-3: [<if output> <forloop> <ifcodition> <else output> else] Wrong answer
#         [<if output> <if conditions> <else> <else out> <forloop>] Carrect answer
```

```python
import random
l = [random.randint(1,100) for i in range(5)]
out = []
for i in l:
    if i % 2 == 0:
        out.append(f'{i}_even')
    else:
        out.append(f'{i}_odd')
out
```

```python
import random
l = [random.randint(1,100) for i in range(5)]
out = [f'{i}_even' if i % 2 == 0 else f'{i}_odd' for i in l]
out
```

**Case-4: if-elif-else with for loop**

```python
# case-1: [<output> <forloop>]
# case-2: [<output> <forloop> <ifcodition>]
# case-3: [<if output> <forloop> <ifcodition> <else output> else] Wrong answer
#         [<if output> <if conditions> <else> <else out> <forloop>] Carrect answer
# case-4: [<if output> <if conditions> <elif conditions> <elif output> <else> <else output> <forloop>] Wrong Answer
#         [[<if output> <if conditions> else<elif out> if<elif conditions> <else> <else output> <forloop>] 
```

```python
# case-1: [<output> <forloop>]
# case-2: [<output> <forloop> <ifcodition>]
# case-3: [<if output> <forloop> <ifcodition> <else output> else] Wrong answer
#         [<if output> <if conditions> <else> <else out> <forloop>] Carrect answer
# case-4: [<if output> <if conditions> <elif conditions> <elif output> <else> <else output> <forloop>] Wrong Answer
#         [<if output> <if conditions> else<elif out> if<elif conditions> <else> <else output> <forloop>] Correct Answer
#         [<if output> <if conditions> else<elif1 out> if<elif1 conditions> else<elif2 out> if<elif2 conditions> <else> <else output> <forloop>] 
```

```python
l = [71,45,63,98]
out = []
for i in l:
    if i >= 75:
        out.append('A')
    elif i >= 60:
        out.append('B')
    elif i >= 35:
        out.append('C')
    else:
        out.append('Fail')
out
```

```python
l = [71,45,63,98]
out = ['A' if i >= 75 else 'B' if i >= 60 else 'C' if i >= 35 else 'Fail' for i in l]
out
```

```python
l = [71,45,63,98,25]
out = ['A' if i >= 75 else 'B' if i >= 60 else 'C' if i >= 35 else 'Fail' for i in l]
out
```
