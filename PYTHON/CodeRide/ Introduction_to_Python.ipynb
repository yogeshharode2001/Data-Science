**Q-17) Before we dive into the exercise, let's understand what reserved keywords actually are. Think of Python as a language with its own special words that have specific jobs. These words are called "reserved keywords" or "keywords" for short. They're reserved because you can't use them for your own variable names or function names - Python has already claimed them for special purposes.
For example, words like if, else, for, and while are keywords because Python uses them to control how your program runs. If Python let you create a variable called if, it would get confused about whether you mean the word "if" for decision-making or your variable named "if". To avoid this confusion, these words are off-limits for regular use.
Python has a built-in module called keyword that keeps track of all these reserved words. This module has a handy list that shows you every keyword currently in the Python language. As Python evolves, keywords can be added or removed, so using this module ensures you always get the current list.
Now, let's work on a program that displays all these reserved keywords so you can see exactly which words are special in Python.**

**Write a Python program that shows all the reserved keywords in the Python language
You'll need to use Python's built-in keyword module to access this list
The program should import the keyword module and then display all keywords
No user input is needed - the keywords are already stored inside Python
This helps you understand which words you cannot use as variable names in your programs**

```python
import keyword
print(keyword.kwlist)
```

**Q-18) Let's build on what we learned from the previous exercise. You now know that Python has reserved keywords stored in the keyword module. But what if you want to know exactly how many keywords exist in Python? That's where counting comes in.
In Python, when you have a list of items, you can easily find out how many items are in that list using a built-in function called len(). The word "len" is short for "length." Just like you can count how many books are on a shelf or how many students are in a classroom, len() counts how many items are in a list.
Since keyword.kwlist gives us a list of all Python keywords, we can simply pass that list to the len() function, and it will tell us the total count. This is a fundamental skill in programming - being able to work with collections of data and get information about them.
Now let's create a program that counts and displays the total number of Python keywords.**


**Write a Python program that counts how many reserved keywords exist in Python**

Use the len() function to count how many keywords are in the list
