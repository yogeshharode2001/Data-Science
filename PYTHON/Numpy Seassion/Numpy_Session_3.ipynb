**1. NumPy Random Functions**

- 1. np.random.random()
    - Generates random decimal numbers between 0 and 1.

```python
import numpy as np
a = np.random.random(5)
print(a)
```

- 2. np.random.rand()
    - Generates random numbers between 0 and 1 using the given dimensions.

```python
a = np.random.rand(3)
print(a)
```

**For a 2D array:**

```python
print(np.random.rand(2, 3))
```

- 3. np.random.randn()
    - Generates random numbers from a standard normal distribution (mean = 0, standard deviation = 1).

```python
a = np.random.randn(5)
print(a)
```

- 4. np.random.uniform()
    - Generates random decimal numbers within a specified range.

**Syntax:**

  - np.random.uniform(start, end, size)

```python
a = np.random.uniform(10, 20, 5)
print(a)
```

- 5. np.random.normal()
    - Generates random numbers following a normal distribution.

**Syntax:**

  - np.random.normal(mean, standard_deviation, size)

```python
a = np.random.normal(50, 10, 5)
print(a)
```

- 6. np.random.choice()
    - Selects random elements from a given list or array.

```python
l = [10, 20, 30, 40, 50]
a = np.random.choice(l)
print(a)
```

**Select multiple random values:**

```python
l = [10, 20, 30, 40, 50]
a = np.random.choice(l, 3)
print(a)
```

- 7. np.random.shuffle()
    - Randomly changes the order of array elements.

**Important: shuffle() changes the original array.**

```python
l = [10, 20, 30, 40, 50]
a = np.array(l)
np.random.shuffle(a)
print(a)
```

```python
a = np.array([10, 20, 30, 40, 50])
np.random.shuffle(a)
print(a)
```

- 8. np.random.seed()
    - Controls random number generation so that you can get the same random results again.

```python
np.random.seed(10)
a = np.random.rand(5)
print(a)
```

```python
np.random.seed(10)
a = np.random.rand(5)
print(a)
```

**1. Mathematical Functions**

- 1. np.sum()
    - Calculates the sum of all elements.

```python
a = np.array([10, 20, 30])
print(np.sum(a))
```

- 2. np.min()
    - Finds the minimum value.

```python
l = [10, 20, 5, 30]
print(np.min(l))
```

- 3. np.max()
    - Finds the maximum value.

```python
l = [10, 20, 5, 30]
print(np.max(l))
```

- 4. np.mean()
    - Calculates the average.

```python
l = [10, 20, 30, 40, 50]
print(np.mean(l))
```

- 5. np.median()
    - Finds the middle value.

```python
l = [10, 20, 30, 40, 50]
print(np.median(l))
```

- 6. np.sqrt()
    - Calculates the square root.

```python
l = [4, 9, 16, 25, 36, 49]
print(np.sqrt(l))
```

- 7. np.square()
    - Calculates the square of each element.

```python
l = [2, 3, 4, 5, 6, 7]
print(np.square(l))
```

- 8. np.power()
    - Raises elements to a specified power.

```python
l = [2, 3, 4, 5, 6, 7]
x = 2
print(np.power(l, x))
```

- 9. np.abs()
    - Returns absolute values (removes negative signs).

```python
l = [-10, -20, 30, -40, 50, -60]
print(np.abs(l))
```

- 10. np.round()
    - Rounds decimal values.

```python
l = [2.345, 3.678, 4.789, 5.432, 6.543, 7.654]
x = 2
print(np.round(l, x))
```

- 11. np.floor()
    - Rounds values down to the nearest integer.

```python
l = [2.8, 3.2, 4.9, 5.8, 6.4, 7.2]
print(np.floor(l))
```

- 12. np.ceil()
    - Rounds values up to the nearest integer.

```python
l = [2.8, 3.2, 4.9, 5.8, 6.4, 7.2]
print(np.ceil(l))
```

- 13. np.exp()
    - Calculates the exponential value (e raised to the power of each element).

```python
l = [1, 2, 3, 4, 5]
print(np.exp(l))
```

**2. Sorting and Searching Functions**

- 14. np.sort()
    - Sorts array elements in ascending order.

```python
l = [40, 10, 30, 20, 60, 50]
print(np.sort(l))
```

- 15. np.argmax()
    - Returns the index of the maximum value.

```python
l = [10, 50, 30, 20, 10, 40]
print(np.argmax(l))
```

- 16. np.argsort()
    - Returns the indices that would sort the array.

```python
l = [10, 50, 30, 20, 60, 40]
print(np.argsort(l))
```

**3. Matrix Functions**

- 17. np.dot()
    - Calculates the dot product of arrays.

```python
a = np.array([1, 2])
b = np.array([3, 4])
print(np.dot(a, b))
```

- 18. np.matmul()
    - Performs matrix multiplication.

```python
a = np.array([[1, 2], [3, 4]])
b = np.array([[5, 6], [7, 8]])
print(np.matmul(a, b))
```

**4. Array Manipulation Functions**

- 19. np.clip()
    - Limits values within a specified range.

```python
l = [10, 20, 30, 40]
a = 15
b = 35
print(np.clip(l, a, b))
```

```python
print(np.clip([10, 20, 30, 40], 15, 35))
```
