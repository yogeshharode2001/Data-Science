**2D Array**

```python
l1=[1,2,3,4]
l2=[10,20,30,40]

# np.array(object, dtype=None)
# np.array(l1,l2)  # l1 ===> object  l2 ===> dtype

l=[l1,l2] # Combining
arr2=np.array(l)
print(arr2)
print(arr2.ndim)  # 2 dime
print(arr2.shape) # (2,4)
print(arr2.size)  # 2*4=8 elements
```

```python
l1=[1,2,3,4]
l2=[10,20,30,40]
l3=[100,200,300,400]

# np.array(object, dtype=None)
# np.array(l1,l2)  # l1 ===> object  l2 ===> dtype

l=[l1,l2,l3] # Combining
arr2=np.array(l)
print(arr2)
print(arr2.ndim)  # 2 dime
print(arr2.shape) # (2,4)
print(arr2.size)  # 2*4=8 elements
```
